```python
import math as m
l2 = 35
l3 = 30
l4 = 43

a = m.pi/2
b = m.pi/3
c = 4*m.pi/9

bx =  -8 + l2*m.cos(a) 
by = 16 + l2*m.sin(a)

cx = bx + l3*m.cos(b)
cy = by - l3*m.sin(b)

dx = cx + l4*m.cos(c)
dy = cy + l4*m.sin(c)

(dx,dy)
```
