```python
import os
import pprint
import json
import statistics as s
import copy as c

path = 'Desktop\Anand\_26apr22'
files = os.listdir(path)
files.sort()
files
```

```python
def readfile(path,file):
    txt_file = open((os.path.join(path,file)) , 'r')
    line = 'a'
    raw_data = []
    while len(line) > 0:
        line = txt_file.readline()
        if len(line) > 0:
            elements = line.split()
#             print(elements) for debugging only
            raw_data.append(elements)
    txt_file.close()
    return raw_data
```

```python
def observations(raw_data):
    keys = ('control', 'peg_2' , 'peg_5', 'peg_8', 'inh_50', 'inh_130', 'inh_200', 'pg2_inh', 'pg5_inh', 'pg8_inh')
    observation = dict.fromkeys(keys)
    
    observation['control'] = raw_data[5][0:5]
    observation['pg2_inh'] = raw_data[5][5:]
    observation['peg_2'] = raw_data[6][0:5]
    observation['pg5_inh'] = raw_data[6][5:]
    observation['peg_5'] = raw_data[7][0:5]
    observation['pg8_inh'] = raw_data[7][5:]
    observation['peg_8'] = raw_data[8]
    observation['inh_50'] = raw_data[9]
    observation['inh_130'] = raw_data[10]
    observation['inh_200'] = raw_data[11]
#     print(observation)
    for i in observation:
        for j in range(len(observation[i])):
            observation[i][j] = float(observation[i][j])

    return observation
```

```python
all_readings = dict.fromkeys(files)
for i in files:
    raw_data = readfile(path,i)
    all_readings[i] = observations(raw_data)
all_readings_json = json.dumps(all_readings, indent = 4) 
# print(all_readings_json)
```

```python
def average_observation(all_readings):
    average_readings = c.deepcopy(all_readings)
    for i in all_readings:
        for j in all_readings[i]:
#             print(average_readings[i]) #for debugging
            average_readings[i][j] = s.mean(average_readings[i][j])
    return average_readings

def max_observation(all_readings):
    max_readings = c.deepcopy(all_readings)
    for i in all_readings:
        for j in all_readings[i]:
#             print(max_readings[i]) #for debugging
            max_readings[i][j] = max(max_readings[i][j])
    return max_readings
```

```python
with open('all_readings.json', 'w', encoding='utf-8') as f:
    json.dump(all_readings, f, ensure_ascii=False, indent=4)

    
average_readings = average_observation(all_readings)
with open('average_readings.json', 'w', encoding='utf-8') as f:
    json.dump(average_readings, f, ensure_ascii=False, indent=4)

    
max_readings = max_observation(all_readings)
with open('max_readings.json', 'w', encoding='utf-8') as f:
    json.dump(max_readings, f, ensure_ascii=False, indent=4)
```
