```python
# tempreature = 300K
Vt = 0.02569
Jo = 0.001
Ron = 0.0005
```

```python
def Jnet_r(V,Ron):
    num = V - Vt*m.log(Vt/(Jo*Ron))
    den = Ron
    J_net = num/den
    return J_net
def Jnet(V):
    J_net = Jo*(m.exp(V/Vt)-1)
    return J_net
```

```python
value = np.linspace(0,6,5000)
v = [round(i,3) for i in value]
Jvalues_r = [Jnet_r(i,0.0005) for i in v]
Jvalues = [Jnet(i) for i in v]
```

```python
# plt.figure(figsize=(5,8))
plt.plot(v,Jvalues_r,label = '$R_{on}$ = $5*10^-4$')
plt.plot(v,Jvalues,label = '$R_{on}$ = 0')
# plt.axhline(y=0, linestyle = '--', color = 'red',label = '$J_{net}$ = 1')
# plt.axvline(x=Vt*m.log(1/Jo + 1) +Ron, linestyle = '--', color = 'red',label = 'V at $J_{net}$ = 1')
plt.grid(visible=None, which='major')
plt.yscale('log')
plt.ylim(0.0001,1000000)
# plt.xlim(0,2)
plt.legend()
plt.title(r'J_Net vs Voltage. $J_o = 10^{-3}$', fontdict = font_title)
plt.ylabel('$log(J_{net}$) (A)',fontdict = font)
plt.xlabel('Voltage (V)',fontdict = font)
```
