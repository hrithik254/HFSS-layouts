# Indium Antimonide Semiconductors

```python
import math as m
import numpy as np
from matplotlib import pyplot as plt
import matplotlib
```

```python
font = {'family': 'serif',
        'color':  'darkred',
        'weight': 'normal',
        'size': 18,
        }
font_s = {'family': 'serif',
        'weight': 'normal',
        'size': 15,
        }
font_title = {'family': 'serif',
        'weight': 'normal',
        'size': 18,
        }
```

```python
#intrinsic carrier concentration
#mass of electron
m_e = 9.1093837*pow(10,-31) #kg
#charge of electron
e = 1.60217663*pow(10,-19) #coulomb 
#mass of electron
mn = 0.013*m_e
mp = 0.18*m_e
#boltzmann constant
k = 1.380649*pow(10,-23) #joule per kelvin or kg m2 s-2 K
k_ev = k/e
#plancks consant
h = 6.62607015*pow(10,-34) #joule second - kg m2 s-1
h_ev = h/e
h_bar = h/(2*m.pi)
#energy gap
eg_ev = 0.22
eg = 0.22*e
```

```python
ni(1/300)
```

```python
def Nc(T):
    """returns density of states in SI units (per cm cube)"""
    N_c = 2*pow((mn*k*T)/(2*m.pi*pow(h_bar,2)),1.5)
    N_c = N_c*pow(10,-6) # m cube to cm cube conversion
    return N_c
def Nv(T):
    """returns density of states in SI units (per cm cube)"""
    N_v = 2*pow((mp*k*T)/(2*m.pi*pow(h_bar,2)),1.5)
    N_v = N_v*pow(10,-6) # m cube to cm cube conversion
    return N_v
```

```python
def ni(inverse_T):
    """ni in per cm cube as Nc and Nv are in per cm cube"""
    n_i = pow(Nc(1/inverse_T)*Nv(1/inverse_T),0.5)*m.exp(-eg*inverse_T/(2*k))
    return n_i
```

```python
ni(1/300)
```

```python
points = np.linspace(0.0002,0.1,50)
ar_T = np.linspace(500,10,50)
carrier_conc = [ni(i) for i in points]
list(zip(points,ar_T))
```

```python
plt.figure(figsize=(10,7))
plt.plot(points,carrier_conc, marker = 'o')
plt.title(r'Intrinsic Carrier Concentration of InSb vs $\frac{1}{T}$', fontdict = font_title)
plt.ylabel('Intrinsic Concentration ($cm^-3$)',fontdict = font)
plt.xlabel(r'$\frac{1}{T}$',fontdict = font)
plt.yscale('log')
plt.xticks(ticks = [0,0.02667,0.053155,0.0776,0.1],
           labels = [r'$\frac{1}{500}$',r'$\frac{1}{370}$',
                     r'$\frac{1}{240}$',r'$\frac{1}{120}$',r'$\frac{1}{10}$']
          ,fontdict = font_s)
plt.grid(visible=None, which='major', axis='x')
```
