```python
from selenium import webdriver
from bs4 import BeautifulSoup
import pandas as pd
```

```python
# To configure webdriver to use Chrome browser, we have to set the path to chromedriver
driver = webdriver.Chrome("/usr/lib/chromium-browser/chromedriver")
```

```python
page_urls = ['https://www.fundoodata.com/citiesindustry/9/3/list-of-electrical-electronics-batteries-companies-in-gurgaon-gurugram?&pageno=1&tot_rows=140&total_results=140&no_of_offices=0'
             ,'https://www.fundoodata.com/citiesindustry/9/3/list-of-electrical-electronics-batteries-companies-in-gurgaon-gurugram?&pageno=2&tot_rows=140&total_results=140&no_of_offices=0'
            ,'https://www.fundoodata.com/citiesindustry/9/3/list-of-electrical-electronics-batteries-companies-in-gurgaon-gurugram?&pageno=3&tot_rows=140&total_results=140&no_of_offices=0'
            ,'https://www.fundoodata.com/citiesindustry/9/3/list-of-electrical-electronics-batteries-companies-in-gurgaon-gurugram?&pageno=4&tot_rows=140&total_results=140&no_of_offices=0'
            ,'https://www.fundoodata.com/citiesindustry/9/3/list-of-electrical-electronics-batteries-companies-in-gurgaon-gurugram?&pageno=5&tot_rows=140&total_results=140&no_of_offices=0']
company_names    = [] #List to store name of the company   
company_urls    = [] #List to store company profile url
for page_url in page_urls: 
    driver.get(page_url)
    content = driver.page_source
    soup = BeautifulSoup(content)
    for a in soup.findAll('div', attrs={'class':'heading'}):
        company     = a.find('a',href = True)
        company_names.append(company.text)
        company_urls.append(company.get('href'))
        
# Using the above code we have found the names of companies and their profile urls, using profile url we are going to extract address and phone
```

```python
for company_url in company_urls[1]:
    driver.get(company_url)
    content = driver.page_source
    page_elements = BeautifulSoup(content)
    for element in page_elements.findAll('div', attrs={'class':'search-page-right-pannel'}):
        print(element)
```
