# Cell Charger (BQ25306)thermsistor calculations 

```python
# the above below is verified with datasheet example as given below
# ---------------------------------------
# vt1 = 0.735
# vt3 = 0.4725
#rhot = 0 (optionl, rhot shifts  HOT temperature T3 up and only slightly shift up COLD temperature T1 )
# rntc_t1 = 27280-rhot
# rntc_t3 = 4910-rhot
# result : rt1 = 4.527 k, rt2 = 23.26 k.
# -------------------------------------
vt1 = 0.735
vt3 = 0.4725
rhot = 0
rntc_t1 = 32554-rhot
rntc_t3 = 1070-rhot

rt2 = (rntc_t1*rntc_t3*(1/vt3 - 1/vt1))/(rntc_t1*(1/vt1 - 1)-rntc_t3*(1/vt3 -1))
rt1 = (1/vt1 - 1)/(1/rt2 + 1/rntc_t1)

print("rt2 = ",rt2,"rt1=",rt1)



```

# Nickel strip thickness calculator



```python
import math
#p changes with Tempeture (T)

def p(T):
  return 5.9*pow(10,-11)*T + 2.76*pow(10,-8)

D = 8580 # material density SI 
h = 0.15*pow(10,-3) # thickness SI
# w = 0.8*pow(10,-3) # material width SI
L = 0.55*pow(10,-3) # material length in meters
I = 20 # current SI
# p = 2.874*pow(10,-8) # electrical resistivity SI
t = 1 # time SI
c = 423 # specific heat in J/Kg⁰C

melting_temp = 1400 # degree celcius
ambient_temp = 45 # degree celcius

pm = p(melting_temp)

w = math.sqrt((I*I*pm*t)/(D*h*h*c*(melting_temp-ambient_temp)))

print("width =",w*1000,"mm")
```
