```python
import math as t
import sympy
from matplotlib import pyplot as plt
import numpy as np
```

```python
#energy relation is 
h = 1.05457182*pow(10,-34)  #reduced planks constant
m = 9.1093837*pow(10,-31)
pi = 3.14159265359
def E(n,L, m = 9.1093837*pow(10,-31)):
    return (pow(h,2)*pow(pi,2)*(2*n+1))/(2*m*pow(L,2))*6.242*pow(10,18)
```

```python
#If we observe the n values, the values used are according to energy level
# degenerate enrgy levels are given the same n.
# we need to account for it in our calculation.
band_gap_calc = []
band_gap_exp = [6.79,4.14,3.2,2.78,1.84]
```

```python
# for benzene L = 2.49 Angstorm, band gap experimental  = 6.79 eV
# we will calculate from n = 2 to n=1 
band_gap1 = E(1,2.49*pow(10,-10))
band_gap_calc.append(band_gap1)
print(band_gap1)
```

```python
# for nahpthalene L = 4.813 Angstorm Bnad Gap = 4.14 eV
# n = 5 to n = 6
band_gap2 = E(4,4.813*pow(10,-10))
band_gap_calc.append(band_gap2)
print(band_gap2)
```

```python
# for anthracene L = 7.278 Angstorm Bnad Gap = 3.2 eV
# n = 5 to n = 6
band_gap3 = E(4,7.278*pow(10,-10))
band_gap_calc.append(band_gap3)
print(band_gap3)
```

```python
# for tetracene L = 9.68 Angstorm Band Gap = 2.78 eV
# n = 5 to n = 6
band_gap4 = E(7,9.68*pow(10,-10))
band_gap_calc.append(band_gap4)
print(band_gap4)
```

```python
# for pentacene L = 12.02 Angstorm Band Gap = 1.84 eV
# n = 5 to n = 6
band_gap5 = E(8,12.02*pow(10,-10))
band_gap_calc.append(band_gap5)
print(band_gap5)
```

```python
plt.plot(band_gap_calc, marker = 'o', label = 'calculated')
plt.plot(band_gap_exp, color = 'Red', marker = 'o', label = 'experimental')
plt.xlabel("Acenes")
plt.ylabel("Bandgap (eV)")
plt.xticks(np.arange(5), ['Benzene', 'Naphthalene', 'Anthracene','Tetracene','Pentacene'])
plt.legend()
```

```python
# c(x) for the whole rod
def S(x_coord,kvalue,w):
    return (1+(kvalue-1)*t.exp(-kvalue*x_coord/w))
```

```python
def C(x_coord,kvalue,L,w):
    if(x_coord <= L-w):
#         print('a')
        return S(x_coord,0.5,w)
        
    elif(x_coord > L-w):
#         print('b')
        return (S(L-w,kvalue,w)/w)*pow((-x_coord+w)/L,kvalue-1)
    
```
