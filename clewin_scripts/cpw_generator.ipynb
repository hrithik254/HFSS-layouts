```python
def write_cpw_cif(gnd_width, slot_width, sig_width, length,
                  metal_layer=1, slot_layer=2,
                  filename="cpw.cif",
                  unit_scale=100):
    """
    Generates a CIF file for a CPW structure.

    Inputs:
    - All dimensions in microns (float allowed, e.g., 24.5)
    - unit_scale = 1000 → converts µm to nm
    """

    def to_int(x):
        return int(x * 100)

    def rect(x1, y1, x2, y2):
        x1i, y1i = to_int(x1), to_int(y1)
        x2i, y2i = to_int(x2), to_int(y2)

        width  = x2i - x1i
        height = y2i - y1i
        cx     = (x1i + x2i) // 2
        cy     = (y1i + y2i) // 2

        return f"B {width} {height} {cx} {cy};\n"

    with open(filename, "w") as f:
        f.write("(Made by Hrithik Singla);\n")
        f.write("(MDHS LAb, CeNSE, IISc Bengaluru);\n")
        f.write("""(Layer names);\n
                L L0; (CleWin: 0 Outline/06808000 06808000);\n
                L L1; (CleWin: 1 Au/0f02e8fe 0f02e8fe);\n
                L L2; (CleWin: 2 dielet_align/0f800080 0f800080);\n
                L L3; (CleWin: 3 gaps/0fb9bcc6 0fb9bcc6);\n
                L L4; (CleWin: 4 markings/0f1383f7 0f1383f7);\n
                L L5; (CleWin: 5 pillars/0f51ee36 0f51ee36);\n
                L L6; (CleWin: 6 wafer_align/0f0413f9 0f0413f9);\n""")
        f.write("DS1 1 1;\n")
        f.write("9 MainSymbol;\n")

        # Metal layer
        f.write(f"L L{metal_layer};\n")

        # Signal
        f.write(rect(-sig_width/2, 0, sig_width/2, length))

        # Grounds
        f.write(rect(-sig_width/2 - slot_width - gnd_width, 0,
                     -sig_width/2 - slot_width, length))
        f.write(rect(sig_width/2 + slot_width, 0,
                     sig_width/2 + slot_width + gnd_width, length))

        # Slot layer
        f.write(f"L L{slot_layer};\n")

        f.write(rect(-sig_width/2 - slot_width, 0,
                     -sig_width/2, length))
        f.write(rect(sig_width/2, 0,
                     sig_width/2 + slot_width, length))

        f.write("DF;\n")
        f.write("C 1;\n")
        f.write("E;\n")

    print(f"✅ CIF file '{filename}' created successfully.")
```
