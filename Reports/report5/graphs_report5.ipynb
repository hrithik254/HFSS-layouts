```python
import os
import pandas as pd
import numpy as np


def clean_header_name(header):
    """
    Cleans header name by removing common constant phrases.
    Modify this function depending on your export format.
    """
    # Remove common phrases
    header = header.replace("Terminal ", "")
    header = header.replace("Parameter Plot1", "")
    header = header.strip()
    return header


def load_csv_directory(directory_path):
    """
    Reads all CSV files in directory.

    Returns:
        data_dict

    Structure:
        data_dict = {
            'csv_filename_without_ext': {
                'frequency': freq_array,
                'variations': {
                    'variation_name1': [values],
                    'variation_name2': [values],
                    ...
                }
            }
        }
    """

    data_dict = {}

    for file in os.listdir(directory_path):
        if file.lower().endswith(".csv"):

            full_path = os.path.join(directory_path, file)
            df = pd.read_csv(full_path)

            # First column assumed to be frequency
            freq = df.iloc[:, 0].values

            variations = {}

            for col in df.columns[1:]:
                clean_name = clean_header_name(col)
                variations[clean_name] = df[col].values.tolist()

            file_key = os.path.splitext(file)[0]

            data_dict[file_key] = {
                "frequency": freq,
                "variations": variations
            }

    return data_dict
```

```python
import matplotlib.pyplot as plt


def plot_variations(data_dict,
                    file_key,
                    fmin=None,
                    fmax=None,
                    title=None):

    freq = data_dict[file_key]["frequency"]
    variations = data_dict[file_key]["variations"]

    freq = np.array(freq)

    # Frequency mask
    if fmin is not None and fmax is not None:
        mask = (freq >= fmin) & (freq <= fmax)
    else:
        mask = np.ones_like(freq, dtype=bool)

    plt.figure(figsize=(10, 6))

    for name, values in variations.items():
        values = np.array(values)
        plt.plot(freq[mask] / 1e9, values[mask], label=name)

    if title is None:
        plt.title(file_key)
    else:
        plt.title(title)

    plt.xlabel("Frequency (GHz)")
    plt.ylabel(file_key)
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
```

```python
directory_path = "."
data = load_csv_directory(directory_path)
```

```python
plot_variations(
    data,
    file_key="Terminal S11 Parameter Plot1",
    fmin=26e9,
    fmax=50e9,
    title="S11 Pillar_r Variations"
)
```
