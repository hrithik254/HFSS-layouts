```python
import csv
import matplotlib.pyplot as plt
```

```python
filename = 'Terminal S11 Parameter Plot1.csv'
fields = []  # Column names
rows = []    # Data rows
freq = []
S11_1um = []
S11_3um = []
S11_5um = []
S11_7um = []
S11_9um = []
S11_10um = []
with open(filename, 'r') as csvfile:
    csvreader = csv.reader(csvfile)  # Reader object

    fields = next(csvreader)  # Read header
    for row in csvreader:     # Read rows
        rows.append(row)

```

```python
for r in rows:
    freq.append(float(r[0]))
    S11_1um.append(float(r[1]))
    S11_3um.append(float(r[2]))
    S11_5um.append(float(r[3]))
    S11_7um.append(float(r[4]))
    S11_9um.append(float(r[5]))
    S11_10um.append(float(r[6]))
```

```python
plt.figure(figsize=(10, 6))
plt.plot(freq[250:], S11_1um[250:], label="S11 1um")
plt.plot(freq[250:], S11_3um[250:], label="S11 3um")
plt.plot(freq[250:], S11_5um[250:], label="S11 5um")
plt.plot(freq[250:], S11_7um[250:], label="S11 7um")
plt.plot(freq[250:], S11_9um[250:], label="S11 9um")
plt.plot(freq[250:], S11_10um[250:], label="S11 10um")
plt.xlabel("Frequency (GHz)")
plt.ylabel("S11 (dB)")
plt.title("S11 Pillar_r Variations")
plt.legend()
plt.grid(True)
plt.show()
```
