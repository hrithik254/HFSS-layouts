```python
import csv
import matplotlib.pyplot as plt

filename = 'through_line_S11.csv'
fields1 = []  # Column names
rows1 = []    # Data rows
freq1 = []
S11_through = []
with open(filename, 'r') as csvfile:
    csvreader = csv.reader(csvfile)  # Reader object

    fields1 = next(csvreader)  # Read header
    for value in csvreader:     # Read rows
        rows1.append(value)
for value in rows1:
    freq1.append(float(value[0]))
    S11_through.append(float(value[1]))

filename = 'transition_S11.csv'
fields2 = []  # Column names
rows2 = []    # Data rows
S11_transition = []
with open(filename, 'r') as csvfile:
    csvreader = csv.reader(csvfile)  # Reader object

    fields2 = next(csvreader)  # Read header
    for value in csvreader:     # Read rows
        rows2.append(value)
for value in rows2:
    S11_transition.append(float(value[1]))
```

```python
plt.plot(freq1, S11_through, label='S11 Through Line')
plt.plot(freq1, S11_transition, label='S11 Transition')
plt.xlabel('Frequency (GHz)')
plt.ylabel('S11 (dB)')
plt.title("comparison of S11 for through line, transition and compensation structure")
plt.grid()
plt.legend()
plt.show()
```
